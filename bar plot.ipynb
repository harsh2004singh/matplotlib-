```python
x= ["python","c","c+","java","js"]
y= [98,75,70,80,82]
plt.xlabel("programming languages")
plt.ylabel("no of users")
plt.title("Bar plot 2D")
c =["r","b","y","m","g"]
plt.bar(x,y,width=0.5,color= c, edgecolor ="r",linewidth =5,linestyle =":")
plt.show()
```

```python
x= ["python","c","c+","java","js"]
y= [98,75,70,80,82]
plt.xlabel("programming languages")
plt.ylabel("no of users")
plt.title("Bar plot 2D")
c =["r","b","y","m","g"]
plt.bar(x,y,width=0.5,color= c, edgecolor ="r",linewidth =5,linestyle =":",alpha =0.5)
plt.show()
```

```python
x= ["python","c","c+","java","js"]
y= [98,75,70,80,82]
plt.xlabel("programming languages")
plt.ylabel("no of users")
plt.title("Bar plot 2D")
c =["r","b","y","m","g"]
plt.bar(x,y,width=0.5,color= c,label ="popularity")
plt.legend()
plt.show()
```

```python
x= ["python","c","c+","java","js"]
y= [98,75,70,80,82]
z =[70,60,61,77,52]
plt.xlabel("programming languages")
plt.ylabel("no of users")
plt.title("Bar plot 2D")
plt.bar(x,y,width=0.5,label ="popularity")
plt.bar(x,z,width=0.5,label ="popularity1")
plt.legend()
plt.show()
```

```python
import numpy as np
x= ["python","c","c+","java","js"]
y= [98,75,70,80,82]
z =[70,60,61,77,52]
width=0.2
p= np.arange(len(x))
p1 =[j+width for j in p]
plt.xlabel("programming languages",fontsize =10)
plt.ylabel("no of users",fontsize =10)
plt.title("Bar plot of programming",fontsize= 15)

plt.bar(p,y,width,label ="popularity in 2026",color ="r")
plt.bar(p1,z,width,label ="popularity in 2007",color ="y")
plt.xticks(p+width/2,x,rotation =10)
plt.legend()
plt.show()
```
