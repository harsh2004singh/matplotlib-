```python
import matplotlib.pyplot as plt
```

```python
x =[1,2,3,4,5,6,7,8]
y= [1,2,3,4,5,6,7,8]
plt.plot(x,y)
plt.show()
```

```python
x =[1,2,3,4,5,6,7,8]
y= [1,2,3,4,5,6,7,8]
plt.plot(x,y,color ='r')
plt.show()
```

```python
x =[1,2,3,4,5,6,7,8]
y= [1,2,3,4,5,6,7,8]
plt.plot(x,y,color ='r')
plt.title('python')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
x =[1,2,3,4,5,6,7,8]
y= [1,2,3,4,5,6,7,8]
plt.plot(x,y,color ='r')
plt.fill_between(x,y)
plt.title('python')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
x =[1,2,3,4,5,6,7,8]
y= [1,2,3,4,5,6,7,8]
plt.plot(x,y,color ='r')
plt.fill_between(x=[2,6],y1=2,y2=6)
plt.title('python')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
x =[1,2,3,4,5,6,7,8]
y= [1,2,3,4,5,6,7,8]
plt.plot(x,y,color ='r')
plt.fill_between(x,y,color ='g')
plt.title('python')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python

x =[1,2,3,4,5,6,7,8]
y= [1,2,3,4,5,6,7,8]
plt.plot(x,y,color ='r')
plt.fill_between(x,y,where=(x>2)&(x<6))
plt.title('python')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
import numpy as np
x =np.array([1,2,3,4,5,6,7,8])
y= np.array([1,2,3,4,5,6,7,8])
plt.plot(x,y,color ='r')
plt.fill_between(x,y,where=(x>2)&(x<6))
plt.title('python')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```
