```python
import matplotlib.pyplot as plt

x = [70, 87, 68, 78, 79, 47, 53, 96, 69, 50]

y = ['Russia', 'China', 'USA', 'France', 'UK',
     'Vietnam', 'UAE', 'Saudi Arabia', 'Iran', 'Israel']

plt.pie(x, labels=y, radius=1)

centre_circle = plt.Circle((0, 0), 0.5, fc='white')
fig = plt.gcf()
fig.gca().add_artist(centre_circle)

plt.title("Country Age")
plt.show()
```
