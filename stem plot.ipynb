```python
import matplotlib.pyplot as plt
import numpy as np
import random
```

```python
x = np.random.randint(-10,10,(10))
y = np.random.randint(-10,10,(10))
```

```python
print(x)
print(y)
```

```python
plt.stem(x,y)
plt.show()
```

```python
plt.stem(x,y,linefmt= ":")
plt.show()
```

```python
plt.stem(x,y,linefmt= ":",markerfmt="r+")
plt.show()
```

```python
plt.stem(x,y,linefmt= ":",markerfmt="r+",bottom=1)
plt.show()
```

```python
plt.stem(x,y,linefmt= ":",markerfmt="r+",bottom=1, basefmt="g")
plt.show()
```

```python
plt.stem(x,y,linefmt= ":",markerfmt="r+",bottom=1,label='stem ')
plt.legend()
plt.show()
```

```python
plt.stem(x,y,linefmt= ":",markerfmt="r+",bottom=1,label='stem ', use_line_collection=False)
plt.legend()
plt.show()
```

```python
plt.stem(x,y,linefmt= ":",markerfmt="r+",bottom=1,label='stem ', use_line_collection=False ,orientation="hori")
plt.legend()
plt.show()
```
