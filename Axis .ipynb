```python
import matplotlib.pyplot as plt
```

```python
x =[1,2,3,4,5]
y =[3,1,5,2,4]

plt.plot(x,y)
plt.show()
```

```python
x =[1,2,3,4,5]
y =[3,1,5,2,4]

plt.plot(x,y)
plt.xticks(x)
plt.show()
```

```python
x =[1,2,3,4,5]
y =[3,1,5,2,4]

plt.plot(x,y)
plt.xticks(x)
plt.yticks(x)
plt.show()
```

```python
x =[1,2,3,4,5]
y =[3,1,5,2,4]

plt.plot(x,y)
plt.xticks(x,labels=['py','c','c++','j','js'])
plt.yticks(x)
plt.show()
```

```python
x =[1,2,3,4,5]
y =[3,1,5,2,4]

plt.plot(x,y)
plt.xlim(0,10)
plt.ylim(0,10)
plt.show()
```

```python
x =[1,2,3,4,5]
y =[3,1,5,2,4]

plt.plot(x,y)
plt.axis([0,10,0,7])
plt.show()
```

# working with text in matplotlib

```python
plt.plot(x,y)
plt.text(2,5,"hello this is harsh",fontsize =15,style ="italic",bbox ={"facecolor":"r"})
plt.annotate("pyhon",xy=(3,2),xytext=(4,4),arrowprops=dict(facecolor='black',shrink=100))
plt.show()
```
